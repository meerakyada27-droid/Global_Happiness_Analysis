```python
# OOP encapsulation
class Happiness:
    def __init__(self, country, score):
        self.country = country
        self.__score = score

    def display(self):
        print("Country:", self.country)
        print("Happiness Score:", self.__score)


country1 = Happiness("India", 4.0)

country1.display()
```

```python
#OOP inheritance
class Country:
    def show_country(self):
        print("Country: India")


class Happiness(Country):
    def show_score(self):
        print("Happiness Score: 4.0")


country1 = Happiness()

country1.show_country()
country1.show_score()
```

```python
#OOP Method Overriding
class Country:
    def show_score(self):
        print("Country Happiness Score")


class India(Country):
    def show_score(self):
        print("India Happiness Score: 4.0")


country1 = India()
country1.show_score()
```

```python
# OOP Super
class Country:
    def __init__(self, country):
        self.country = country


class Happiness(Country):
    def __init__(self, country, score):
        super().__init__(country)
        self.score = score

    def display(self):
        print("Country:", self.country)
        print("Happiness Score:", self.score)


country1 = Happiness("India", 4.0)
country1.display()
```

```python
# OOP Abstraction
from abc import ABC, abstractmethod

class Country(ABC):
    @abstractmethod
    def show_score(self):
        pass


class Happiness(Country):
    def show_score(self):
        print("India Happiness Score: 4.0")


country1 = Happiness()
country1.show_score()
```
